```python
# Dictionary to store performance results
model_performance = {}
```

```python
# 2. Initialize the models with a specified random state for consistency
models = {
    'Logistic Regression': LogisticRegression(max_iter=200, random_state=42),
    'SVM': SVC(random_state=42),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=40, random_state=42)
}
```

```python
# Initialize Logistic Regression model
log_reg = LogisticRegression(max_iter=200, random_state=42)
```

```python
# Calculate 5-fold cross validation score
log_reg_scores = cross_val_score(log_reg, X, y, cv=5)
log_reg_mean = np.mean(log_reg_scores)
```

```python
# Store and print the performance
model_performance['Logistic Regression'] = log_reg_mean
print(f"Logistic Regression Average Score: {log_reg_mean:.4f}")
```

```python
# Initialize SVM model
svm_model = SVC(random_state=42)
```

```python
# Calculate 5-fold cross validation score
svm_scores = cross_val_score(svm_model, X, y, cv=5)
svm_mean = np.mean(svm_scores)
```

```python
# Store and print the performance
model_performance['SVM'] = svm_mean
print(f"SVM Average Score: {svm_mean:.4f}")
```

```python
# Initialize Decision Tree model
dec_tree = DecisionTreeClassifier(random_state=42)
```

```python
# Calculate 5-fold cross validation score
dec_tree_scores = cross_val_score(dec_tree, X, y, cv=5)
dec_tree_mean = np.mean(dec_tree_scores)
```

```python
# Store and print the performance
model_performance['Decision Tree'] = dec_tree_mean
print(f"Decision Tree Average Score: {dec_tree_mean:.4f}")
```

```python
# Initialize Random Forest model
rand_forest = RandomForestClassifier(n_estimators=40, random_state=42)
```

```python
# Calculate 5-fold cross validation score
rand_forest_scores = cross_val_score(rand_forest, X, y, cv=5)
rand_forest_mean = np.mean(rand_forest_scores)
```

```python
# Store and print the performance
model_performance['Random Forest'] = rand_forest_mean
print(f"Random Forest Average Score: {rand_forest_mean:.4f}")
```
